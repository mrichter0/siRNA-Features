```python
import glob, os, multiprocessing as mp, pandas as pd, pymol2

session_file   = '2.pse'
reference_obj  = 'ENSG00000051382_unmod'
gro_dir        = '/data/home/mrichte3/RNASeq/unmod/step4/'
csv_path       = 'unmod_log2fc.csv'
output_csv     = 'unmod_log2fc_with_dist.csv'
ncpus          = 24

gro_files = sorted(glob.glob(os.path.join(gro_dir, '*.gro')))
if not gro_files:
    raise FileNotFoundError('No .gro files found in directory!')

def distance_worker(gro_path):
    comp_base = os.path.splitext(os.path.basename(gro_path))[0]
    comp_name = f'{comp_base}_unmod'
    try:
        with pymol2.PyMOL() as pm:
            pm.cmd.reinitialize()
            pm.cmd.load(session_file)
            pm.cmd.hide('cartoon', 'all')
            pm.cmd.hide('ribbon',  'all')

            obj = 'tmp_obj'
            pm.cmd.load(gro_path, object=obj)
            pm.cmd.remove(f'{obj} and solvent')
            pm.cmd.remove(f'{obj} and (resn NA or resn Na or resn Cl or resn CL or resn Na+)')
            pm.cmd.align(obj, reference_obj)

            sel1 = f'first ({obj} and resi 6 and name N6)'
            sel2 = f'first ({reference_obj} and resi 6 and name N6)'
            dist = pm.cmd.get_distance(sel1, sel2)

        # print(f'[INFO] {comp_name}: distance = {dist:.3f} Å')
        return comp_name, dist
    except Exception as e:
        print(f'[WARN] {comp_name}: {e}')
        return comp_name, float('nan')

with mp.Pool(processes=ncpus) as pool:
    results = pool.map(distance_worker, gro_files)

dist_df = pd.DataFrame(results, columns=['comp_name', 'resi6_N6'])
orig_df = pd.read_csv(csv_path)
merged  = orig_df.merge(dist_df, on='comp_name', how='left')
merged.to_csv(output_csv, index=False)

print(f'[DONE] Distances added → {output_csv}')
```

```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr
import matplotlib
matplotlib.use('Agg')            # ensures PNG rendering without a GUI
import matplotlib.pyplot as plt

csv_path      = 'unmod_log2fc_with_dist.csv'   # merged CSV from previous step
plot_filename = 'log2fc_vs_distance.png'

# ────────── load & prepare ──────────
df = pd.read_csv(csv_path)
df['abs_log2FC'] = df['log2FC'].abs()
clean = df.dropna(subset=['abs_log2FC', 'resi6_N6'])

# ────────── correlations ──────────
pearson_corr, pearson_p   = pearsonr(clean['abs_log2FC'], clean['resi6_N6'])
spearman_corr, spearman_p = spearmanr(clean['abs_log2FC'], clean['resi6_N6'])

print(f'Pearson r  = {pearson_corr:.4f}  (p = {pearson_p:.4g})')
print(f'Spearman ρ = {spearman_corr:.4f}  (p = {spearman_p:.4g})')

# ────────── scatter-plot ──────────
plt.figure(figsize=(6,5))
plt.scatter(clean['abs_log2FC'], clean['resi6_N6'], alpha=0.7, edgecolor='k')
plt.xlabel('|log2FC|')
plt.ylabel('resi6_N6 distance (Å)')
plt.title('Correlation: |log2FC| vs. resi6_N6 distance')
plt.grid(True, linestyle='--', linewidth=0.5, alpha=0.6)
plt.tight_layout()
plt.savefig(plot_filename, dpi=300)
print(f'Plot saved → {plot_filename}')
```
