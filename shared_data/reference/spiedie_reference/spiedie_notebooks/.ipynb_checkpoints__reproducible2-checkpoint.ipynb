```python
import subprocess
import time
import os
os.environ['PATH'] = '/data/home/mrichte3/gromacs-2024.2/install/bin:' + os.environ['PATH']
os.environ['LD_LIBRARY_PATH'] = '/data/home/mrichte3/gromacs-2024.2/install/lib:' + os.environ.get('LD_LIBRARY_PATH', '')
os.environ['GMX_MAXBACKUP'] = '-1'
equi_prefix = "step4.0_minimization"
prod_prefix = "step5_production"
prod_step = "step5cpu"
def run_command(command, input_text=None):
    """Run a shell command with optional input and capture output."""
    result = subprocess.run(command, capture_output=True, text=True, input=input_text)
    # print(result.stdout)
    # print(result.stderr)
    return result


nsteps = 100
while nsteps <= 10100:
    print(f"Running simulation with nsteps = {nsteps}")
    overall_start_time = time.time()
    
    for i in range(1, 11):
        with open(f"{prod_prefix}.mdp", 'r') as file:
            mdp_content = file.readlines()

        with open(f"{prod_prefix}_temp2.mdp", 'w') as file:
            for line in mdp_content:
                if line.startswith("nsteps"):
                    line = f"nsteps = {nsteps}\n"
                file.write(line)

        grompp_command = [
            "gmx", "grompp", "-f", f"{prod_prefix}_temp2.mdp", "-o", f"{prod_step}.tpr",
            "-c", f"{equi_prefix}.gro", "-p", "topol.top", "-n", "index.ndx", "-maxwarn", "2"
        ]
        run_command(grompp_command)

        mdrun_command = [
            "gmx", "mdrun", "-v", "-deffnm", prod_step, "-ntmpi", "1", "-ntomp", "32", "-nb", "cpu"
        ]
        run_command(mdrun_command)

        mv_command = ["mv", "step5cpu.gro", f"step5cpu_{i}.gro"]
        run_command(mv_command)

    overall_end_time = time.time()
    overall_elapsed_time = overall_end_time - overall_start_time
    print(f"Total simulation time for nsteps={nsteps}: {overall_elapsed_time:.2f} seconds")

    %run graphs_cpu.py
    nsteps += 100
```
